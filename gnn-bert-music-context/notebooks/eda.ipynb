# EDA — MagnaTagATune
Quick look at the dataset before training: tag frequency, split sizes, and audio availability.

```python
import json
import pandas as pd
import matplotlib.pyplot as plt

mtat = pd.read_csv('../data/processed/mtat_master.csv')
meta = json.load(open('../data/processed/dataset_meta.json'))
TOP50 = meta['top50_tags']

print(mtat.shape)
mtat.head()
```

## Split sizes (all clips vs. audio-having clips)

```python
print('All clips:\n', mtat.split_folder.value_counts())
print('\nAudio-having clips:\n', mtat[mtat.has_audio].split_folder.value_counts())
```

## Top-50 tag frequency (training split)

```python
freq = mtat[mtat.split_folder == 'train'][TOP50].sum().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(14, 4))
ax.bar(range(len(freq)), freq.values, color='#4a7ba7')
ax.set_xticks(range(len(freq))); ax.set_xticklabels(freq.index, rotation=80, ha='right', fontsize=7)
ax.set_ylabel('training count'); ax.set_title('Top-50 tag frequency')
plt.tight_layout(); plt.show()
```

## Caption length distribution

```python
mtat['caption_len'] = mtat.caption.fillna('').str.split().apply(len)
mtat.caption_len.hist(bins=30)
plt.xlabel('words per caption'); plt.ylabel('count'); plt.title('Caption length')
plt.show()
```
